### a. Create a list of numbers and print all the even numbers

```python
# Sample list defined directly in the program
numbers = [18, 7, 28, 53, 48, 86, 19, 60, 93, 100]

# Filter even numbers using list comprehension
even_numbers = [num for num in numbers if num % 2 == 0]

# Output the results
print("Sample List: ", numbers)
print("Even Numbers:", even_numbers)
```

### b. Create a string and print it in reverse order. Now reverse the words also.

```python
# Get input from the user
word = input("Enter a single word: ")

# Reverse the word using slicing
reversed_word = word[::-1]

# Print the reversed word
print("Reversed word is :", reversed_word)
```

```python
# Get a sentence from the user
sentence = input("Enter a sentence: ")

# Option 1: Reverse every character in the sentence
reversed_characters = sentence[::-1]

# Option 2: Reverse the order of the words
reversed_words = " ".join(sentence.split()[::-1])

# Display the results
print("\nCharacter-by-character reversed sentence:")
print(reversed_characters)

print("\nWord-order reversed sentence:")
print(reversed_words)
```
