# Assignment 4: Word Cloud Generator
This notebook processes text, removes stopwords, calculates word frequencies, generates a Word Cloud, and saves it in `.jpg` format for evaluation.

```python
# Install wordcloud & nltk if needed (uncomment if running in a new environment)
# !pip install wordcloud nltk matplotlib
```

```python
# Import necessary libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# Ensure necessary NLTK data is downloaded
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

### Step 1: Input Text
Paste the response you get from ChatGPT describing yourself, your skills, and your passions into the `text` variable below.

```python
# Provided text: Replace the placeholder text below with your ChatGPT output
text = """
I am a passionate software developer and AI enthusiast exploring artificial intelligence, machine learning, and data science.
My technical skills include Python programming, deep learning, computer vision, data analysis, neural networks, and problem solving.
I enjoy building intelligent applications, designing clean algorithms, working with modern technologies, and continuously learning new concepts.
Curiosity, innovation, collaboration, critical thinking, creativity, and dedication define my approach to technology and software engineering.
"""

print("Provided Text:")
print(text.strip())
```

### Step 2: Tokenization and Stopword Removal
Convert words to lowercase, tokenize them, and filter out common stopwords as well as punctuation.

```python
# Tokenization of words (convert to lowercase for uniformity)
words = word_tokenize(text.lower())
print("Total tokens:", len(words))

# Remove stopwords and non-alphabetical characters
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word.isalpha() and word not in stop_words]

print("Filtered words count:", len(filtered_words))
print("Filtered words sample:", filtered_words)
```

### Step 3: Frequency Distribution
Count the frequency of each unique word.

```python
# Frequency distribution of words
word_freq = Counter(filtered_words)
print("Word Frequencies:")
print(word_freq)
```

### Step 4: Generate WordCloud & Save in JPG Format
Generate the word cloud, display it with Matplotlib, and save it in `.jpg` format as required for assignment evaluation.

```python
# Create a WordCloud
wordcloud = WordCloud(width=800, height=400, background_color='white').generate_from_frequencies(word_freq)

# Plotting the word cloud
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')  # Hide axes
plt.title("My Word Cloud", fontsize=14)
plt.show()

# Save the word cloud in JPG format (mandatory for evaluation)
wc_image = wordcloud.to_image()
wc_image.convert('RGB').save('wordcloud.jpg', 'JPEG')
print("Successfully saved word cloud as 'wordcloud.jpg'!")
```
