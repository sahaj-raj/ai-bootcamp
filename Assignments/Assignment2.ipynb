```python
import pandas as pd

data = {
    "Name": [" Aadi  ", "Aarav..", " Nistha "],
    "Age": [19, 20, None],
    "Marks": [87, 93, 77]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Clean names using RegEx
df["Name"] = df["Name"].str.replace(r"[^A-Za-z\s]", "", regex=True)
df["Name"] = df["Name"].str.replace(r"\s+", " ", regex=True)
df["Name"] = df["Name"].str.strip()

# Fill missing age
df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\nCleaned Data:")
print(df)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)

y = np.sin(x)

plt.plot(x, y)

plt.xlabel("x (radians)")
plt.ylabel("sin(x)")
plt.title("Sine Wave")

plt.grid(True)

plt.show()
```
