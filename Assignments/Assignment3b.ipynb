Write Python code to demonstrate application of Linear Regression in the electronics industry.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
voltage = np.array([5, 6, 7, 8, 9, 10, 11, 12]).reshape(-1, 1)

power = np.array([10, 12, 14, 16, 18, 21, 23, 25])

print("Voltage:", voltage.flatten())
print("Power:", power)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    voltage,
    power,
    test_size=0.25,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
model = LinearRegression()

model.fit(X_train, y_train)

print("Model trained!")
```

```python
y_pred = model.predict(X_test)

print("Actual Power:", y_test)
print("Predicted Power:", y_pred)
```

```python
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```

```python
new_voltage = np.array([[15]])

predicted_power = model.predict(new_voltage)

print("Predicted power at 15V:", predicted_power[0], "Watts")
```

```python
plt.figure(figsize=(7, 4))

plt.scatter(voltage, power, label="Actual Data")
plt.plot(voltage, model.predict(voltage), label="Regression Line")

plt.xlabel("Voltage (V)")
plt.ylabel("Power Consumption (W)")
plt.title("Linear Regression: Voltage vs Power")
plt.legend()

plt.show()
```
