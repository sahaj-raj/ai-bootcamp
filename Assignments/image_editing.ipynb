```python
import cv2
import matplotlib.pyplot as plt

image_path = r"C:\Users\HP\Downloads\download343.jpg"
image = cv2.imread(image_path)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```

```python
height, width = image.shape[:2]
cropped_image = image[height // 4:3 * height // 4, width // 4:3 * width // 4]
cropped_image_rgb = cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB)

plt.imshow(cropped_image_rgb)
plt.axis("off")
plt.show()
```

```python
reversed_image = cv2.flip(image, 1)
reversed_image_rgb = cv2.cvtColor(reversed_image, cv2.COLOR_BGR2RGB)

plt.imshow(reversed_image_rgb)
plt.axis("off")
plt.show()
```
